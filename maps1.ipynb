```python
from ipyleaflet import Map, Marker
from geopy import Nominatim
```

```python
# get the mapCenter coordinates

locator = Nominatim(user_agent="myGeocoder")
location = locator.geocode("Osnabrück, Germany")

mapCenter = (location.latitude, location.longitude)
mapCenterMarker = Marker(location=mapCenter, draggable=False)
```

```python
m = Map(center=mapCenter, zoom=10, scroll_wheel_zoom=True)
m.add_layer(mapCenterMarker);
display(m);
```
